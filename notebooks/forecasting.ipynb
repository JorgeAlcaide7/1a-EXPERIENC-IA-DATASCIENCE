```python
# Cargar datos de inferencia desde data/raw/inferencia
import pandas as pd
import os

# Path relativo desde la carpeta notebooks
inferencia_path = os.path.join('..','data','raw','inferencia','ventas_2025_inferencia.csv')

# Cargar el archivo csv en dataframe
inferencia_df = pd.read_csv(inferencia_path)

# Mostrar información del dataframe
print("--- inferencia_df (primeras 5 filas) ---")
print(inferencia_df.head())
print(f"\nDimensiones: {inferencia_df.shape[0]} filas, {inferencia_df.shape[1]} columnas")
print(f"\nColumnas: {inferencia_df.columns.tolist()}")
```

```python
# 🤖 Transformar inferencia_df aplicando el mismo pipeline que df_processed

import pandas as pd
import numpy as np
import holidays as hol
from pathlib import Path

# 1. CONVERTIR FECHA A DATETIME
inferencia_df['fecha'] = pd.to_datetime(inferencia_df['fecha'])
print(f"✓ Fecha convertida: {inferencia_df['fecha'].dtype}")
print(f"  Rango: {inferencia_df['fecha'].min()} a {inferencia_df['fecha'].max()}")

# 2. DATOS DE COMPETENCIA
# El fichero de inferencia 2025 ya incluye las columnas de competencia
# (Amazon, Decathlon, Deporvillage). El histórico competencia.csv solo cubre
# 2021-2024, así que mergear para 2025 solo añadiría columnas vacías y
# duplicaría las existentes (sufijos _x/_y). Solo mergeamos si faltan.
comp_cols = ['Amazon', 'Decathlon', 'Deporvillage']
if not all(c in inferencia_df.columns for c in comp_cols):
    competencia_path = Path('..') / 'data' / 'raw' / 'entrenamiento' / 'competencia.csv'
    competencia = pd.read_csv(competencia_path)
    competencia['fecha'] = pd.to_datetime(competencia['fecha'])
    inferencia_df = inferencia_df.merge(
        competencia[['fecha', 'producto_id'] + comp_cols],
        how='left',
        on=['fecha', 'producto_id']
    )
print(f"✓ Datos de competencia disponibles: {inferencia_df.shape}")

# 3. VARIABLES TEMPORALES Y DE CALENDARIO
spanish_holidays = hol.Spain()

# Variables básicas de fecha
inferencia_df['año'] = inferencia_df['fecha'].dt.year
inferencia_df['mes'] = inferencia_df['fecha'].dt.month
inferencia_df['dia_mes'] = inferencia_df['fecha'].dt.day
inferencia_df['dia_semana'] = inferencia_df['fecha'].dt.day_name()
inferencia_df['dia_semana_num'] = inferencia_df['fecha'].dt.dayofweek

# Variables derivadas
inferencia_df['trimestre'] = inferencia_df['fecha'].dt.quarter
inferencia_df['semana_ano'] = inferencia_df['fecha'].dt.isocalendar().week
inferencia_df['num_dias_mes'] = inferencia_df['fecha'].dt.daysinmonth

# Indicadores
inferencia_df['es_fin_semana'] = inferencia_df['dia_semana_num'].isin([5, 6]).astype(int)
inferencia_df['es_festivo'] = inferencia_df['fecha'].apply(lambda x: 1 if x in spanish_holidays else 0)

# Black Friday: último viernes de noviembre
inferencia_df['es_Black_Friday'] = 0
for year in inferencia_df['año'].unique():
    noviembre = pd.date_range(start=f'{year}-11-01', end=f'{year}-11-30', freq='D')
    black_friday_date = [d.date() for d in noviembre if d.weekday() == 4][-1]
    inferencia_df.loc[inferencia_df['fecha'].dt.date == black_friday_date, 'es_Black_Friday'] = 1

# Cyber Monday: primer lunes después del Black Friday
inferencia_df['es_Cyber_Monday'] = 0
for year in inferencia_df['año'].unique():
    noviembre = pd.date_range(start=f'{year}-11-01', end=f'{year}-11-30', freq='D')
    black_friday_date = [d for d in noviembre if d.weekday() == 4][-1]
    cyber_monday_date = black_friday_date + pd.Timedelta(days=3)
    inferencia_df.loc[inferencia_df['fecha'].dt.date == cyber_monday_date.date(), 'es_Cyber_Monday'] = 1

# Eventos especiales
inferencia_df['es_Navidad'] = ((inferencia_df['mes'] == 12) & (inferencia_df['dia_mes'].isin([24, 25, 26]))).astype(int)
inferencia_df['es_Ano_Nuevo'] = ((inferencia_df['mes'] == 1) & (inferencia_df['dia_mes'].isin([1, 2, 3]))).astype(int)
inferencia_df['es_Verano'] = inferencia_df['mes'].isin([7, 8]).astype(int)

# Indicador de puente
inferencia_df['es_puente'] = 0
for idx, row in inferencia_df.iterrows():
    fecha = row['fecha']
    dia_anterior = fecha - pd.Timedelta(days=1)
    dia_siguiente = fecha + pd.Timedelta(days=1)
    if (fecha.weekday() == 4 and dia_siguiente in spanish_holidays) or \
       (fecha.weekday() == 0 and dia_anterior in spanish_holidays):
        inferencia_df.at[idx, 'es_puente'] = 1

# Variables de distancia temporal
inferencia_df['dias_hasta_fin_mes'] = inferencia_df['num_dias_mes'] - inferencia_df['dia_mes']
inferencia_df['dias_hasta_fin_trimestre'] = inferencia_df.apply(
    lambda row: (pd.Timestamp(year=row['año'], month=(row['trimestre']*3), day=1) + 
                 pd.offsets.MonthEnd(0) - row['fecha']).days, axis=1
)
inferencia_df['dias_hasta_fin_ano'] = inferencia_df.apply(
    lambda row: (pd.Timestamp(year=row['año'], month=12, day=31) - row['fecha']).days, axis=1
)

print(f"✓ Variables temporales creadas: {inferencia_df.shape}")

# 4. VARIABLES DE DESCUENTO Y PRECIO (ANTES DE LAGS)
# Descuento porcentaje
inferencia_df['descuento_porcentaje'] = ((inferencia_df['precio_venta'] - inferencia_df['precio_base']) / 
                                          inferencia_df['precio_base']) * 100

# Precio competencia y ratio
inferencia_df['precio_competencia'] = inferencia_df[['Amazon', 'Decathlon', 'Deporvillage']].mean(axis=1)
inferencia_df['ratio_precio'] = inferencia_df['precio_venta'] / inferencia_df['precio_competencia']

# Eliminar columnas de competidores
inferencia_df = inferencia_df.drop(columns=['Amazon', 'Decathlon', 'Deporvillage'])
print(f"✓ Variables de precio creadas: {inferencia_df.shape}")

# 5. VARIABLES LAG Y MEDIAS MÓVILES
# IMPORTANTE: Para inferencia necesitamos los datos históricos de 2024
# Cargar df_processed para obtener los últimos días de datos históricos
df_processed_path = Path('..') / 'data' / 'processed' / 'df_processed.csv'
df_historico = pd.read_csv(df_processed_path)
df_historico['fecha'] = pd.to_datetime(df_historico['fecha'])

# Concatenar histórico con inferencia para calcular lags correctamente
df_combinado = pd.concat([df_historico, inferencia_df], ignore_index=True).sort_values(['producto_id', 'fecha'])

# Crear lags y media móvil por producto
for lag in range(1, 8):
    df_combinado[f'unidades_vendidas_lag{lag}'] = df_combinado.groupby('producto_id')['unidades_vendidas'].shift(lag)

df_combinado['unidades_vendidas_ma7'] = df_combinado.groupby('producto_id')['unidades_vendidas'].transform(
    lambda x: x.rolling(window=7, min_periods=1).mean().shift(1)
)

# Filtrar solo las filas de inferencia (año 2025)
inferencia_df = df_combinado[df_combinado['año'] == 2025].copy()
print(f"✓ Variables lag creadas: {inferencia_df.shape}")

# 6. ONE-HOT ENCODING
# Crear copias con sufijo _h
inferencia_df['nombre_h'] = inferencia_df['nombre']
inferencia_df['categoria_h'] = inferencia_df['categoria']
inferencia_df['subcategoria_h'] = inferencia_df['subcategoria']

# Aplicar OHE
ohe_cols = ['nombre_h', 'categoria_h', 'subcategoria_h']
df_ohe = pd.get_dummies(
    inferencia_df[ohe_cols].astype(str), 
    prefix=['nombre_h', 'categoria_h', 'subcategoria_h'], 
    dtype=int
)

# Concatenar columnas OHE
inferencia_df = pd.concat([inferencia_df, df_ohe], axis=1)
print(f"✓ One-Hot Encoding aplicado: {df_ohe.shape[1]} columnas añadidas")

# 7. ASEGURAR QUE TIENE TODAS LAS COLUMNAS DE df_processed
# Cargar las columnas de df_processed para asegurar compatibilidad
df_proc_cols = df_historico.columns.tolist()
missing_cols = set(df_proc_cols) - set(inferencia_df.columns)

# Añadir columnas faltantes con valor 0
for col in missing_cols:
    if col not in ['unidades_vendidas', 'ingresos']:  # No añadir target
        inferencia_df[col] = 0

# Reordenar columnas para que coincidan (excepto target si no existe)
final_cols = [c for c in df_proc_cols if c in inferencia_df.columns]
inferencia_df = inferencia_df[final_cols]

print(f"✓ Estructura alineada con df_processed")

# 8. FILTRAR SOLO NOVIEMBRE (eliminar octubre)
inferencia_df = inferencia_df[inferencia_df['mes'] == 11].copy()
print(f"✓ Filtrado a noviembre: {inferencia_df.shape[0]} registros")

# 9. GUARDAR ARCHIVO TRANSFORMADO
output_path = Path('..') / 'data' / 'processed'
output_path.mkdir(parents=True, exist_ok=True)
inferencia_df.to_csv(output_path / 'inferencia_df_transformado.csv', index=False)

print(f"\n✅ Transformación completada!")
print(f"   Archivo guardado: {output_path / 'inferencia_df_transformado.csv'}")
print(f"   Dimensiones finales: {inferencia_df.shape}")
print(f"   Columnas: {len(inferencia_df.columns)}")
print(f"   Rango de fechas: {inferencia_df['fecha'].min()} a {inferencia_df['fecha'].max()}")
```
